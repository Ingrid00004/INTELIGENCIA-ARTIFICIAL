```python
import pandas as pd

tabela = pd.read_csv("base_custos_transporte.csv")
tabela.head()
```

**Etapa 2 – Identificar as variáveis**

a) Qual é a variável independente?
*   mes.

b) Qual é a variável dependente?
*   custo_transporte.

c) O que representa cada variável no contexto do problema?
*   mes: os ultimos 12 meses de dados;
*   custo_transporte: custo total de transporte no mês em reais.

**Etapa 3 – Construir um gráfico**

```python
import matplotlib.pyplot as plt

plt.scatter(tabela["mes"], tabela["custo_transporte"])
plt.xlabel("Mês")
plt.ylabel("Custo de Transporte (R$)")
plt.title("Custo de Transporte por Mês")
plt.show()
```

O gráfico mostra que o custo de transporte foi aumentando ao longo dos meses, e esse crescimento parece bem constante, os pontos praticamente formam uma linha reta subindo.

**Etapa 4 – Encontrar uma função**

```python
import numpy as np

a, b = np.polyfit(tabela["mes"], tabela["custo_transporte"], 1)
print("a:", a)
print("b:", b)
```

A função encontrada
*   f(x) = 1791,61x + 42204,55

**Etapa 5 – Interpretar a função**

a) O que representa o coeficiente a no contexto do problema?
*   Representa o custo de transporte aumentando a cada mês, em média R$1791,61.

b) O que representa o coeficiente b?
*   Representa o custo de transporte estimado no "mês zero".

c) O valor de a indica crescimento ou redução dos custos?
*   Crescimento.

**Etapa 6 – Representar a função graficamente**

```python
import matplotlib.pyplot as plt

plt.scatter(tabela["mes"], tabela["custo_transporte"], label="Dados reais")
plt.plot(tabela["mes"], a * tabela["mes"] + b, color="red", label="Função ajustada")
plt.xlabel("Mês")
plt.ylabel("Custo de Transporte (R$)")
plt.title("Custo de Transporte por Mês com Função Ajustada")
plt.legend()
plt.show()
```

**Etapa 7 – Fazer previsões**

```python
mes_13 = a * 13 + b
mes_14 = a * 14 + b
mes_15 = a * 15 + b

print("Mês 13:", mes_13)
print("Mês 14:", mes_14)
print("Mês 15:", mes_15)
```

**Etapa 9 – Conclusão**

Com base na análise realizada, é possível observar que os custos de transporte vêm crescendo de forma constante ao longo dos 12 meses analisados, a uma taxa média de aproximadamente 1.791,61 por mês.
